## Random Tensor
#### Why Random Tensors
# Random tensors are important because the way many neural network is that they start with tensors full of random numbers and then adjust those random numbers to better represent the data.
# ` Start with random numbers -> look at data -> update random number -> look at data -> update random numbers`

```python
# Create a random tensor of size (3,4)
random_tensor = torch.rand(1,3,4)
random_tensor
```

```python
random_tensor.ndim
```

```python
random_tensor.shape
```
