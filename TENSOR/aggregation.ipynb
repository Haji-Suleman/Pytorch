### Finding the min, max, mean, sum, etc (tensor aggregation)

```python
import torch
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x  = torch.arange(10,101,10)
print(torch.min(x),x.min())
print(torch.max(x),x.max())
print(torch.sum(x),x.sum())
x_floatingPoint = x.type(torch.float16)
print(torch.mean(x_floatingPoint),x.type(torch.float32).mean())
print(torch.median(x))
```
