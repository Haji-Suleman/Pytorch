## Reshaping, stacking, squeezing  and unsqueezing tensors

* Reshaping - reshapes an input tensor  to a defined shape
* View - Return a view of an input tensor of certain shape but keep the same memory as the original tensor
* Stacking -  combine multiple tensors on the top of each other (vstack) or side by side (hstack)

* Squeeze - removes all `1` dimensions from a tensor
* Unsqueeze - add a `1` dimension to a target tensor
* Permute - Return a view of the input with dimensions permuted (swapped) in a certain way

```python
# Let's create a tensor
x  = torch.arange(1.,13.)
x,x.shape 
```

```python
x_Reshaped = x.reshape(6,2)
x_Reshaped, x_Reshaped.shape
```

```python
x_Viewed = x.view(6,2)
x_Viewed,x_Viewed.shape
```

```python
# Changing x_Viewed changes x_Reshaped (because a view of a tensor shares the same memory as the original input)

x_Viewed[:,0] = 5
x,x_Viewed
```

```python
x_stacked = torch.stack([x,x,x])
x_stacked
```

```python
# torch.squeeze()-remove all single dimentions from target tensor
x = torch.tensor([[3, 4, 5]])
print(x.squeeze(dim=1))
```

```python
# torch.unsqueeze - adds a single dimention to a target tensor at a specific dim (dimension) 
```

```python
# permute
x = torch.rand(size=(224, 224, 3))
print(torch.permute(x, (2, 0, 1)).size())
```
